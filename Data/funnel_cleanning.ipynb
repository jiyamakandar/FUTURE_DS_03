```python
oct_df = pd.read_csv(
    r"C:\Users\jiyam\OneDrive\Desktop\Marketing Funnel & Conversion Performance Analysis\2019-Oct.csv",
    nrows=300000
)
```

```python
nov_df = pd.read_csv(
    r"C:\Users\jiyam\OneDrive\Desktop\Marketing Funnel & Conversion Performance Analysis\2019-Nov.csv",
    nrows=300000
)
```

```python
df = pd.concat([oct_df, nov_df], ignore_index=True)
```

```python
funnel_counts = df['event_type'].value_counts()

print(funnel_counts)
```

```python
views = funnel_counts.get('view', 0)
carts = funnel_counts.get('cart', 0)
purchases = funnel_counts.get('purchase', 0)

print("Views:", views)
print("Carts:", carts)
print("Purchases:", purchases)
```

```python
#View to Cart Conversion
view_to_cart = (carts / views) * 100

print("View to Cart Conversion Rate:", round(view_to_cart, 2), "%")
```

```python
#View to Purchase Conversion
view_to_purchase = (purchases / views) * 100

print("View to Purchase Conversion Rate:", round(view_to_purchase, 2), "%")
```

```python
#Drop-Off Rate
dropoff_rate = ((views - purchases) / views) * 100

print("Overall Drop-off Rate:", round(dropoff_rate, 2), "%")
```

```python
#Create Funnel Table
import pandas as pd

funnel_summary = pd.DataFrame({
    'Stage': ['Views', 'Carts', 'Purchases'],
    'Count': [views, carts, purchases]
})

print(funnel_summary)
```

```python
#Create Funnel Chart
import matplotlib.pyplot as plt

stages = ['Views', 'Carts', 'Purchases']
values = [views, carts, purchases]

plt.figure(figsize=(8,5))

plt.bar(stages, values)

plt.title("Marketing Funnel Analysis")
plt.ylabel("Count")

plt.show()
```
